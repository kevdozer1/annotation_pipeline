# BridgeEngine Explore

Five DuckDB queries over the latest local Mode A snapshot.

```python
from bridgeengine.paths import latest_snapshot_id
from bridgeengine.query import demo_queries, preview, run_query

snapshot_id = latest_snapshot_id()
snapshot_id
```

```python
preview(snapshot_id)
```

```python
# Query 1: structured filter on label coverage
run_query(snapshot_id, demo_queries()['mask_coverage']).head()
```

```python
# Query 2: caption content filter
run_query(snapshot_id, demo_queries()['caption_put']).head()
```

```python
# Query 3: captions joined with depth dynamic range
run_query(snapshot_id, demo_queries()['caption_depth_range']).head()
```

```python
# Query 4: episode count by labeler success
run_query(snapshot_id, demo_queries()['labeler_success_counts'])
```

```python
# Query 5: provenance trace for one episode
run_query(snapshot_id, demo_queries()['provenance_trace'])
```
